## KDE and Displot

```python
print('''
KDE (Kernel Density Estimation)

KDE is a smooth version of a histogram.
KDE is a smooth curve that shows the distribution of data.

Instead of bars like a histogram, KDE draws a continuous smooth line.

It shows:

👉 how data is distributed using a smooth curve instead of bars.
''')

sns.displot(df['total_bill'],bins=6,kde=True)

print()
```

## Histplot


```python
sns.histplot(df['total_bill'],bins=5,kde=True)
# plt.xticks(x)
```

## sns Subplots

```python
#subplots() --------------->      plt.subplot(rows, columns, position)
plt.figure(figsize=(10,6))

plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=12,kde=True)

plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=12,kde=True)

print()
fig,ax=plt.subplots(1,2,figsize=(10,6))

sns.histplot(df['tip'],ax=ax[0])
sns.histplot(df['total_bill'],ax=ax[1])


plt.tight_layout()

plt.show()
```

## Joint Plots

```python
print('''
Joint Plot (sns.jointplot())
Definition

A Joint Plot is a Seaborn visualization that shows:

the relationship between two variables
and the distribution of each variable together in one graph.

It combines:

Scatter Plot
Histogram or KDE Plot

into a single visualization.
''')

sns.jointplot(x='total_bill',y='tip',data=df, kind='scatter')

sns.jointplot(x='total_bill',y='tip',data=df, kind='reg')

sns.jointplot(x='total_bill',y='tip',data=df, kind='kde')

sns.jointplot(x='total_bill',y='tip',data=df, kind='hex')
```

```python
# fig,ax=plt.subplots(1,2,figsize=(10,6))
fig, ax = plt.subplots(2,2, figsize=(10,8))

sns.scatterplot(x='total_bill', y='tip', data=df, ax=ax[0,0])

sns.regplot(x='total_bill', y='tip', data=df, ax=ax[0,1])

sns.kdeplot(x='total_bill', y='tip', data=df, ax=ax[1,0])

sns.histplot(df['total_bill'], ax=ax[1,1])

plt.tight_layout()
```

## Pair Plots

```python
#Pair plot

print('''
2. Pair Plot
Definition

Automatically shows:

relationships between ALL numerical columns
distributions for each column

Example:

sns.pairplot(df)

Creates many scatter plots together.

vars    Select specific columns.

hue=Adds category colors.

sns.pairplot(df, hue='sex')

Different colors for:

Male
Female,


diag_kind='kde',
kind='scatter',
palette='cool'


''')
# print(df)

sns.pairplot(df,hue='sex')
plt.show()
print()
sns.pairplot(df[['total_bill','tip','day']], hue='day')
plt.show()
print()

sns.pairplot(
    df,
    vars=['total_bill', 'tip', 'size'],
    hue='size',
    diag_kind='kde',
    kind='scatter'
    ,palette='cool'
)

plt.show()
```

## Stripplot and Swarmplot

### Strip Plot (sns.stripplot())
### Definition

#### A Strip Plot shows individual data points across categories.

#### It helps visualize:

#### distribution
#### spread
#### clustering

#### using scatter points.

#### Syntax
#### sns.stripplot(x= , y= , data= )

#### What It Shows

#### Each dot:
#### 👉 one observation/data point.

#### Example:

#### one customer bill
#### Problem in Stripplot

#### Points can overlap.

#### Many dots may stack on each other.

```python
sns.stripplot(x='day',y='total_bill',data=df)
plt.show()
print()

# jitter helps to spread a data points
sns.stripplot(x='day',y='total_bill',data=df,jitter=0.2)
plt.show()
```

## lmplot (linear model plot)
### sns.lmplot() Parameters

### Definition

### Higher-level regression plot with grouping/faceting support.

### Syntax
### sns.lmplot(x= , y= , data= )

### Important Parameters (lmplot)
### Parameter	Meaning
### x	x variable
### y	y variable
### data	dataframe
### hue	category coloring
### col	separate columns
### row	separate rows
### fit_reg	regression line
### ci	confidence interval
### markers	marker types
### palette	colors
### height	figure height
### aspect	width ratio

```python
sns.lmplot(x='total_bill',y='tip',data=tips)   # we can use hue in this.......
plt.show()
df = sns.load_dataset('tips')

sns.lmplot(
    
    x='total_bill',          # x variable
    y='tip',                 # y variable
    
    data=df,                 # dataframe
    
    hue='sex',               # category coloring
    
    col='smoker',            # separate columns
    row='time',              # separate rows
    
    fit_reg=True,            # show regression line
    
    ci=95,                   # confidence interval
    
    markers=['o', '*'],      # marker styles
    
    palette='Set1',          # colors
    
    height=5,                # figure height
    
    aspect=1.2             # width ratio
    
)

plt.show()
```

### Important Parameters (regplot)
### Parameter	Meaning
### x	x-axis variable
### y	y-axis variable
### data	dataframe
### ci	confidence interval
### scatter	show scatter points
### fit_reg	show regression line
### marker	point shape
### color	overall color
### line_kws	line customization
### scatter_kws	scatter customization
### ax	subplot axes

```python
sns.regplot(x='total_bill',y='tip',data=tips)
plt.show()

sns.regplot(
    x='total_bill',
    y='tip',
    data=df,
    ci=75,
    fit_reg=True,marker='*',scatter=True,line_kws={'color':'red'},scatter_kws={'s':100}
)
plt.show()

# regplot()  ax
fig, ax = plt.subplots(1,2)

sns.regplot(
    x='total_bill',
    y='tip',
    data=df,
    ax=ax[0]
)

sns.regplot(
    x='size',
    y='tip',
    data=df,
    ax=ax[1],
)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import pandas as pd
# import cufflinks as cf
```

```python
import plotly.express as px
df = sns.load_dataset('tips')

fig = px.scatter(
    df,
    x='total_bill',
    y='tip',
    color='sex'
)

fig.show()

fig = px.line(
    df,
    x='size',
    y='total_bill'
)

fig.show()

fig = px.bar(
    df,
    x='day',
    y='total_bill',
    color='sex'
)

fig.show()
fig = px.histogram(
    df,
    x='total_bill',
    nbins=20
)

fig.show()

fig = px.box(
    df,
    x='day',
    y='total_bill',
    color='sex'
)

fig.show()


fig.show()

```

```python
fig = px.violin(
    df,
    x='day',
    y='total_bill',
    color='sex'
)

fig.show()

fig = px.pie(
    df,
    names='day'
)

fig.show()

corr = df.corr(numeric_only=True)

fig = px.imshow(
    corr,
    text_auto=True
)

fig.show()

fig = px.area(
    df,
    x='size',
    y='total_bill'
)

fig.show()

fig = px.density_heatmap(
    df,
    x='total_bill',
    y='tip'
)

fig.show()

fig = px.scatter_matrix(
    df,
    dimensions=['total_bill', 'tip', 'size'],
    color='sex'
)

fig = px.scatter_3d(
    df,
    x='total_bill',
    y='tip',
    z='size',
    color='sex'
)

fig.show()

fig = px.sunburst(
    df,
    path=['day', 'sex']
)

fig.show()

fig = px.treemap(
    df,
    path=['day', 'sex'],
    values='total_bill'
)

fig.show()
```
